# Multi-LLM Debate Framework (3 Models, 7 Iterative Debate Rounds)

This notebook executes the debate workflow matching the architectural diagram:
1. **Prompt / Question (User Input)**
2. **Parallel Input to Model 1, Model 2, and Model 3**
3. **7 Rounds of Iterative Debate** (Models exchange arguments, counter-arguments & refine responses)
4. **Final Output** (Synthesized Consensus Result)

```python
import os
import json
import asyncio
import nest_asyncio
from dotenv import load_dotenv

# Apply nest_asyncio to allow asyncio execution within Jupyter
nest_asyncio.apply()

# Load API Keys
load_dotenv()

print("Environment initialized.")
```

## 1. Import Framework Modules
Importing our modular model manager, execution controller, evaluator, and visualizer.

```python
from models.model_manager import ModelManager
from models.health_check import run_all_health_checks
from execution.query_execution import ExecutionController
from evaluation.evaluator import DebateEvaluator
from evaluation.visualization import DebateVisualizer

print("Modules imported successfully.")
```

## 2. Health Checks (Model 1, Model 2, Model 3)

```python
await run_all_health_checks()
```

## 3. Execute 7-Round Iterative Debate Across All 3 Models
All 3 models generate initial arguments, then perform 7 rounds of cross-examination and refinement.

```python
query = "What is Capital of India?"
max_rounds = 2 # 7 Rounds of Iteration

controller = ExecutionController()

print(f"Prompt / Question: '{query}'")
print(f"Executing {max_rounds} Debate Iteration Rounds across Model 1, Model 2, and Model 3...")

debate_result = await controller.run_full_debate(query, max_rounds=max_rounds)

print("""
=== 7-ROUND DEBATE ITERATION COMPLETED ===""")
```

## 4. Final Output & Consensus Synthesis

```python
evaluator = DebateEvaluator()
metrics = evaluator.evaluate(debate_result)

visualizer = DebateVisualizer()
visualizer.display_metrics(metrics)

print("--- Synthesized / Consensus Result ---")
print(json.dumps(debate_result['final_consensus'], indent=2))
```
